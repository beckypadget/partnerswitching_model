```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_a10_b3 = pd.read_csv("evolution_agg_a10-b3.csv")
df_a3_b3 = pd.read_csv("evolution_agg_a3-b3.csv")
df_a3_b10 = pd.read_csv("evolution_agg_a3-b10.csv")
```

## Highly cooperative population

```python
sns.lineplot(data=df_a10_b3, x='all_uncertainty', y='mean_ps', color="purple")
plt.xlabel("Population information", fontsize=16)
plt.ylabel("Probability of switching", fontsize=16)
plt.title("High", fontsize=18)
plt.savefig(f"prob_switching-a10-b3.png")
plt.show()
```

## Mid cooperative population

```python
sns.lineplot(data=df_a3_b3, x='all_uncertainty', y='mean_ps', color="purple")
plt.xlabel("Population information", fontsize=16)
plt.ylabel("Probability of switching", fontsize=16)
plt.title("Intermediate", fontsize=18)
plt.savefig(f"prob_switching-a3-b3.png")
plt.show()
```

## Low cooperation population

```python
sns.lineplot(data=df_a3_b10, x='all_uncertainty', y='mean_ps', color="purple")
plt.xlabel("Population information", fontsize=16)
plt.ylabel("Probability of switching", fontsize=16)
plt.title("Low", fontsize=18)
plt.savefig(f"prob_switching-a3-b10.png")
plt.show()
```
